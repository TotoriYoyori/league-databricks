```sql
WITH random_matches AS (
    SELECT match_id, ROW_NUMBER() OVER(ORDER BY match_id) AS surr_id
    FROM (
        SELECT DISTINCT match_id
        FROM gold.matchend_player_stats
    )
)

SELECT *
FROM (
    SELECT match_id
    FROM random_matches
    WHERE surr_id % 4 = 0
)
JOIN gold.matchend_player_stats USING(match_id)
;
```
